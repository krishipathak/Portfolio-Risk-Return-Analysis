```python
import pandas as pd
import yfinance as yf
import numpy as np
```

```python
portfolio = ['RELIANCE.NS','TCS.NS','INFY.NS','HDFCBANK.NS','ICICIBANK.NS']
```

```python
start_date = '2023-01-01'
end_date = '2026-01-01'
```

```python
for stock in portfolio:
  yf.download(stock, start=start_date, end= end_date)
```

```python
print(portfolio)
```

```python
raw_data = {}

for stock in portfolio:
  raw_data[stock] = yf.download(stock, start=start_date, end= end_date)
```

```python
raw_data.keys()
```

```python
raw_data.values()
```

```python
def standardize_data(df,ticker):
  df = df.reset_index()
  df.columns = [col[0] for col in df.columns]
  df["ticker"] = ticker
  return df
```

```python
processed_data = {}
for stock in portfolio:
  processed_data[stock] = standardize_data(raw_data[stock], stock)
```

```python
processed_data.values()
```

```python
processed_data.keys()
```

```python
processed_data.values()
```

```python
master_data = pd.concat(processed_data.values(), ignore_index = True)
```

```python
master_data.isnull().sum()
```

```python
master_data.info()
```

```python
master_data.head()
```

```python
master_data = master_data.sort_values(by = ["ticker","Date"])
```

```python
master_data.reset_index(drop = True)
```

```python
master_data = master_data.reset_index(drop = True)
```

```python
master_data.groupby("ticker")
```

```python
master_data.groupby("ticker").groups.keys()
```

```python
master_data["Daily_Return"] = (master_data.groupby("ticker")["Close"].pct_change())
```

```python
master_data.head(15)
```

```python
master_data["Cumulative_Return"] = ((1+master_data["Daily_Return"].groupby(master_data['ticker']).cumprod()))
```

```python
master_data.head()
```

```python
master_data.reset_index(drop= True)
```

```python
portfolio_metadata = pd.DataFrame({'ticker':['RELIANCE.NS','TCS.NS','INFY.NS','HDFCBANK.NS','ICICIBANK.NS'], 'weight':[0.3,0.25,0.2,0.15,0.1], 'Sector': ["Energy","Information Technology","Information Technology","Financial Services","Financial Services"]})
```

```python
portfolio_metadata
```

```python
portfolio_metadata.columns
```

```python
master_data = master_data.merge(portfolio_metadata, on = 'ticker', how = 'left')
```

```python
master_data.head()
```

```python
master_data['Weighted_Return'] = (master_data['Daily_Return']*master_data['weight'])
```

```python
portfolio_returns = (master_data.groupby("Date")["Weighted_Return"].sum().reset_index())
```

```python
portfolio_returns.columns = ["Date", "Portfolio_Daily_Return"]
```

```python
portfolio_returns.head()
```

```python
portfolio_returns["Portfolio_Cumulative_Return"] = ((1+portfolio_returns["Portfolio_Daily_Return"]).cumprod())
```

```python
portfolio_returns.head()
```

```python
portfolio_returns['Portfolio_Cumulative_Percent'] = (portfolio_returns["Portfolio_Cumulative_Return"]-1)
```

```python
benchmark_return = benchmark_data[["Date","Benchmark_Return", "Benchmark_Cumulative_Return"]]
```

```python
benchmark_return
```

```python
benchmark_return.columns = ["Date","Benchmark_Return","Benchmark_Cumulative_Return"]
```

```python
portfolio_comparison = portfolio_returns[["Date","Portfolio_Daily_Return","Portfolio_Cumulative_Return"]].merge(benchmark_return,on = "Date", how = "inner")
```

```python
portfolio_comparison.dropna()
```

```python
beta = portfolio_comparison["Portfolio_Daily_Return"].cov(portfolio_comparison["Benchmark_Return"]/portfolio_comparison["Benchmark_Return"].var())
```

```python
beta
```

```python
years = (benchmark_return["Date"].iloc[-1] - benchmark_return["Date"].iloc[0]).days/365.25
```

```python
years
```

```python
#benchmark CAGR
benchmark_cagr = (benchmark_return['Benchmark_Cumulative_Return'].iloc[-1]/benchmark_return['Benchmark_Cumulative_Return'].iloc[0])**(1/years)-1
```

```python
benchmark_cagr*100
```

```python
portfolio_cagr = (portfolio_returns["Portfolio_Cumulative_Return"].iloc[-1]/portfolio_returns["Portfolio_Cumulative_Return"].iloc[0])**(1/years)-1
```

```python
portfolio_cagr*100
```

```python
required_return = risk_free_rate+beta*(benchmark_cagr-risk_free_rate)
```

```python
alpha = portfolio_cagr - required_return
alpha
```

```python
alpha*100
```

```python
treynor_ratio = (portfolio_cagr - risk_free_rate)/ beta
```

```python
treynor_ratio*100

```

```python
benchmark_treynor = (benchmark_cagr - risk_free_rate)
```

```python
benchmark_treynor *100
```
